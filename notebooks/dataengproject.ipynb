```python
import re

with open("key.js", encoding="utf-8") as f:
    js = f.read()

section = js.split("var measures = {", 1)[1].split("};", 1)[0]

pattern = r'(\d+)\s*:\s*\["([^"]*)",\s*"([^"]*)"\]'
matches = re.findall(pattern, section)

traits_df = pd.DataFrame(matches, columns=["trait_id", "trait_low", "trait_high"])

traits_df["trait_id"] = traits_df["trait_id"].astype(int)
traits_df = traits_df[traits_df["trait_id"].between(1, 525)]

#traits_df["column_name"] = "BAP" + traits_df["trait_id"].astype(str)

traits_df
```

```python
with open("key.js", encoding="utf-8") as f:
    js = f.read()

section = js.split("var subjects =", 1)[1]

pattern = r'"([^"]+)"\s*:\s*\{\s*((?:(?!\n\s*"\w+"\s*:\s*\{).)*)'

characters = []

for work_code, body in re.findall(pattern, section, re.S):
    names = re.findall(
        r'(\d+)\s*:\s*\["([^"]+)",\s*"([^"]+)"\]',
        body
    )

    for char_id, name, image in names:
        characters.append({
            "character_code": f"{work_code}/{char_id}",
            "character_name": name,
            "image": image,
            "work_code": work_code
        })

character_df = pd.DataFrame(characters)
character_df[["character_code","character_name"]]
```

```python
with open("characterScores2125x525.js", "r", encoding="utf-8") as f:
    with open("characterScores.txt", "w", encoding="utf-8") as fw:
        for i in range(2200):
            content = f.readline()
            content = content.replace("[", "")
            content = content.replace("]", "")
            content = content.replace(";", "")
            content = content.replace("{", "")
            content = content.replace("}", "")
            content = content.replace(": ", ",")
    
            

            if(i==0):
                fw.write(content[:-1]+'\n')
                #print(content[:-1])
            else:
                fw.write(content[:-2]+'\n')
                #print(content[:-2])
            
```
